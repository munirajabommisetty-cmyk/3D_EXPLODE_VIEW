```python
import numpy as np
import cv2

# Create a 5 × 5 grayscale image
image = np.zeros((5, 5), dtype=np.uint8)

# Make the center pixel white
image[2, 2] = 255

print(image)

# Enlarge the image so we can see the pixels
large_image = cv2.resize(
    image,
    (500, 500),
    interpolation=cv2.INTER_NEAREST
)

cv2.imshow("My First Image", large_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Open webcam
camera = cv2.VideoCapture(0)

if not camera.isOpened():
    print("Error: Could not open webcam.")
    raise SystemExit

print("Hand detection prototype started.")
print("Press Q to quit.")

while True:
    # Step 1: Capture a frame
    success, frame = camera.read()

    if not success:
        print("Could not read webcam frame.")
        break

    # Step 2: Mirror the frame
    frame = cv2.flip(frame, 1)

    # Step 3: Convert BGR image to HSV
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Step 4: Define a rough skin-color range
    lower_skin = np.array([0, 20, 70], dtype=np.uint8)
    upper_skin = np.array([20, 255, 255], dtype=np.uint8)

    # Step 5: Keep pixels within this color range
    mask = cv2.inRange(hsv, lower_skin, upper_skin)

    # Step 6: Remove some small noise
    kernel = np.ones((5, 5), np.uint8)

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel
    )

    # Step 7: Find connected contours
    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # Step 8: Find the largest region
    if contours:
        largest_contour = max(contours, key=cv2.contourArea)

        area = cv2.contourArea(largest_contour)

        # Ignore very small regions
        if area > 3000:
            x, y, w, h = cv2.boundingRect(largest_contour)

            # Draw bounding box
            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                "Possible hand region",
                (x, max(y - 10, 25)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

    # Step 9: Display original frame and mask
    cv2.imshow("Hand Detection Prototype", frame)
    cv2.imshow("Color Segmentation Mask", mask)

    # Exit when Q is pressed
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

camera.release()
cv2.destroyAllWindows()
```
